# Geometrical flows

Given the type of simulations we are interested in, the software comes together with some implementation of geometrical flows. The available ones are:
- Mean curvature flow as described in [this article](https://www.sciencedirect.com/science/article/pii/S0021999106003615)
- Willmore flow with possible spontaneous curvature as implemented [here](https://www.academia.edu/127649470/Computational_Parametric_Willmore_Flow_with_Spontaneous_Curvature_and_Area_Difference_Elasticity_Effects)

We also introduce a novel implementation that merges the stabilization implemented in [this article](https://epubs.siam.org/doi/10.1137/140982696) together with the algorithm from [Barret, Garke and Nurnberg](https://www.academia.edu/127649470/Computational_Parametric_Willmore_Flow_with_Spontaneous_Curvature_and_Area_Difference_Elasticity_Effects).

Sometimes, the stabilization by itself is still not sufficient to guarantee smooth evolutions of real meshes. This is why we also added the tangential mesh redistribution from [Duan and Li](https://epubs.siam.org/doi/full/10.1137/23M1551857).

In the following notebook we will present how to use these solvers. The main reason these were implemented were as means to describe the motion of some boundaries in the simulation (together with others external factors dictated by the `ADRSolver`).
