
# Advection-Diffusion-Reaction (ADR) Solver

We demonstrate here how to solve general advection-diffusion-reaction (ADR) equations in the form:

\begin{equation*}
\partial_t u + \nabla\cdot(\vec{b}u - d \nabla u) + cu = f
\end{equation*}

where
- $u$ is the species considered
- $\vec{b}$ is the advection field
- $d$ is the diffusion coefficient
- $c$ is the reaction coefficient
- $f$ is the right-hand side

The code gives the opportunity to solve ADR equations
-  in the bulk
-  on surfaces 
-  coupled bulk-surface equations

using a common structure.  

We will go through all the cases and show:
- How to setup the various problems
- How to test that the code is correct
- The functionality available
